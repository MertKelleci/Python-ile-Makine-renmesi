# Q-Learning

- Pekiştirmeli öğrenme alanında kullanılan bir öğrenme tekniğidir.
- Bir ajanın belirli bir ortamda öğrenmesini optimize eder.
- Temelde, bir ajanın, bir ortamda, belirli eylemler gerçekleştirip bu eylemlerin sonuçlarını gözlemler ve en uygun eylemleri öğrenir.
- Model tabanlı değildir, ortam hakkında önceden bilgi sahibi olmak zorunluluğu yoktur.
- Alınabilecek aksiyonları içeren sütunlar ve bu aksiyonlar sonrasında geçilecek durumlardan oluşan bir *Q-Table* oluşturulur.
    - Eğer *Q-Table* yerine bir *Deep Learning* algoritması konulursa *Deep Q-Learning* elde edilmiş olunur. *Q-Table* gerçek hayattaki sistemleri, sistemlerin karmaşıklığı nedeniyle, açıklamakta yetersiz kalabilir.
$$\text{Initialize Q-Table} \rightarrow \text{Choose an Action} \rightarrow \text{Perform the Action} \rightarrow \text{Measure Reward} \rightarrow \text{Update Q-Table}$$

![Q-Learning Formula](https://media.datacamp.com/legacy/v1725960040/image_d200c39908.png)
- $S$: Current state
- $A$: Action taken by the agent
- $S_{t+1}$: The Next state the agent moves to 
- $a$: All possible actions that can be taken from the state $S_{t+1}$
- $R$: The reward recieved for taking action $A$ in state $S$
- $\gamma$: The discount factor which balances immediate reward with future rewards. Keeping it high makes the agent avoid punishment, while lowering it makes it go for the reward but may cause it to not avoid punishment.
- $\alpha$: The learning rate determining how much new information affects the old *Q-Values*
