# Q-Learning

- Pekiştirmeli öğrenme alanında kullanılan bir öğrenme tekniğidir.
- Bir ajanın belirli bir ortamda öğrenmesini optimize eder.
- Temelde, bir ajanın, bir ortamda, belirli eylemler gerçekleştirip bu eylemlerin sonuçlarını gözlemler ve en uygun eylemleri öğrenir.
- Model tabanlı değildir, ortam hakkında önceden bilgi sahibi olmak zorunluluğu yoktur.
    - $S$: Current state
    - $A$: Action taken by the agent
    - $S_{t+1}$: The Next state the agent moves to 
    - $a$: The  best next action in the state $S_{t+1}$
    - $R$: The reward recieved for taking action $A$ in state $S$
    - $\gamma$: The discount factor which balances immediate rewards with future rewards
    - $\alpha$: The learning rate determining how much new information affects the old *Q-Values*
