<h1 style="text-align:center;">Naive Bayes</h1>

$\normalsize P(A|B)=\large\frac{P(A\cap B)}{P(B)}$

$\normalsize P(C_i|X) = \large\frac{P(X|C_i)P(C_i)}{P(X)}$

    &ensp;$P(\text{age <= 30 | buys\_computer = yes}) = 2/9 = 0.222$\

    &ensp;&ensp;&ensp;&ensp;&ensp;&ensp;&ensp;.\
    &ensp;&ensp;&ensp;&ensp;&ensp;&ensp;&ensp;.\
    &ensp;&ensp;&ensp;&ensp;&ensp;&ensp;&ensp;.\
    &ensp;$P(\text{credit\_rating = fair | buys\_computer= yes}) = 6/9 =0.667$\
    &ensp; $P(\text{credit\_rating = fair | buys\_computer= no}) = 2/5 =0.4$
* $\text{X = (age <= 30, income = medium, student = yes, credit\_rating = fair)}:$\
    &ensp;$P(X|C_i): P(X|\text{buys\_computer = yes}) = 0.222*0.444*0.667*0.667=0.044$\
    &ensp;&ensp;&ensp;&ensp;&ensp;&ensp;&ensp;&ensp;&ensp;&ensp;&ensp;$P(X|\text{buys\_computer} = \text{no}) = 0.6*0.4*0.2*0.4=0.019$

    &ensp;&ensp;&ensp;&ensp;&ensp;&ensp;&ensp;&ensp;&ensp;&ensp;&ensp;&ensp;&ensp;&ensp;&ensp;&ensp;&ensp;&ensp;&ensp;$P(X|\text{buys\_computer} = \text{no})*P(\text{buys\_computer} = \text{no}) = 0.007$

*Naive-Bayes* teoremi formulü:
- $P(C_k | X): \text{Posterior}$
- $P(X | C_k): \text{Likelyhood}$
- $P(C_k): \text{Prior}$
$$P(C_k | X) = \frac{P(X | C_k) \cdot P(C_k)}{P(X)}$$
$$\downarrow$$
$$P(C_k) \prod P(X|C_k)$$

$$P(C_k) \rightarrow P(\text{yes}) = 9/14, P(\text{no}) = 5/14$$
---
$$X = (\text{sunny, cool, high, strong})$$
$$P(\text{yes}) \prod P(X|\text{yes})$$
$$\downarrow$$
$$P(\text{yes}) \cdot P(\text{sunny}|\text{yes}) \cdot P(\text{cool}|\text{yes})\cdot P(\text{high}|\text{yes})\cdot P(\text{strong}|\text{yes})$$
$$P(\text{Tennis Played}|\text{sunny, cool, high, strong}) = 5.291*10^{-3}$$
$$P(\text{Tennis Not Played}|\text{sunny, cool, high, strong}) = 2.057*10^{-2}$$

**Değerlerin normalize edilmesi:**

$$P(C_{a|b}|X) = \frac{P(C_{a|b})}{P(C_{a|b}) + P(C_{b|a})}$$
$$\downarrow$$
$$P(\text{yes}|X) \rightarrow \frac{P(\text{yes}|X)}{P(\text{yes}|X) + P(\text{no}|X)} = 0.2045$$
$$P(\text{no}|X) \rightarrow \frac{P(\text{no}|X)}{P(\text{yes}|X) + P(\text{no}|X)} 0.7954$$


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

veriler = pd.read_csv("veriler.csv")

x = veriler.iloc[:,1:4].values
y = veriler.iloc[:,4:].values

from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.33, random_state=0)
```

```python
x_train
```

```python
from sklearn.naive_bayes import GaussianNB
gnb = GaussianNB()
```

```python
import warnings
warnings.filterwarnings("ignore")
```

```python
gnb.fit(x_train, y_train)
```

```python
y_pred = gnb.predict(x_test)
```

```python
from sklearn.metrics import confusion_matrix
```

```python
cm = confusion_matrix(y_true=y_test, y_pred=y_pred)
cm
```
