## Information Gain

Decision Tree oluşturulurken dallanma sırası neye göre belirlenecek(ilk hangi özellikler kontrol edilecek)? Bunu bulmak için, verilerin *entropy* değerleri hesaplanır ve *entropy* değeri en düşük olan seçilir. *Entropy* değeri 0 ile 1 arasındadır, hesaplaması yapılan kriterin miktarı birbirine ne kadar yakınsa değer o kadar yüksek çıkacaktır.

#### Quinlan's ID3 Algorithm

$$\normalsize \textrm{Info}(D) = - \sum\limits_{i=1}^{m}p_i log_2 (p_i)$$

$$\normalsize \textrm{Info}_A (D)=\sum\limits_{j=1}^{v} \frac{|D_j|}{D}\times I(D_j)$$

$$\normalsize \textrm{Gain}(A)=\textrm{Info}(D)-\textrm{Info}_A(D)$$

**CART: Classification and Regression Tree.

$$\text{Gain}(S,A) = \text{Entropy}(S)  - \sum_{v\epsilon \text{Values}(A)} \frac{|S_v|}{|S|}\text{Entropy}(S_v)$$

## Gini Impurity:
- Gini belirsizliği, bir başka karar ağacı algoritması olan CART(Classification and Regression Trees) tarafından kullanılan bir ölçümdür.
- Bir düğümdeki veri noktalarının farklı sınıflara dağılımının ne kadar karışık olduğunu ölçer.
- Gini karışıklığı [0 - 0.5] arasında değer alır (2'li sınıflandırma yapılmıyorsa [0 - 1] arası). 0 en karışık durumu, 0.5 ise en saf durumu temsil eder.
- Bir düğümün Gini karışıklığı ne kadar düşükse, o kadar daha homojen sınıflara sahip olur ve tercih edilir.

$$\text{Gini Impurity} = 1 - \sum p_i^2$$

#### Gini Impurity (Money):
$$\text{Money} \rightarrow (7\text{R}, 3\text{P})$$
$$\text{Money(Poor)} \rightarrow (3\text{C}, 0\text{T}, 0\text{SI}, 0\text{S})$$
$$\text{Gini} = 1 - \left[\left(\frac{3}{3}\right)^2 + \left(\frac{0}{3}\right)^2 + \left(\frac{0}{3}\right)^2 + \left(\frac{0}{3}\right)^2\right] = 0$$
$$\text{Money(Rich)} \rightarrow (3\text{C}, 2\text{T}, 1\text{SI}, 1\text{S})$$
$$\text{Gini} = 1 - \left[\left(\frac{3}{7}\right)^2 + \left(\frac{2}{7}\right)^2 + \left(\frac{1}{7}\right)^2 + \left(\frac{1}{7}\right)^2\right] = 0.6939$$
$$\text{Gini(Money)} \rightarrow 0*\frac{3}{10} + 0.6939*\frac{7}{10} = 0.4857$$


#### Gini Impurity (Parents):
$$\text{Parents} \rightarrow (5\text{Y}, 5\text{N})$$
$$\text{Parents(Yes)} \rightarrow (5\text{C}, 0\text{T}, 0\text{SI}, 0\text{S})$$
$$\text{Gini} = 1 - \left[\left(\frac{5}{5}\right)^2 + \left(\frac{0}{5}\right)^2 + \left(\frac{0}{5}\right)^2 + \left(\frac{0}{5}\right)^2\right] = 0$$
$$\text{Parents(No)} \rightarrow (1\text{C}, 2\text{T}, 1\text{SI}, 1\text{S})$$
$$\text{Gini} = 1 - \left[\left(\frac{1}{5}\right)^2 + \left(\frac{2}{5}\right)^2 + \left(\frac{1}{5}\right)^2 + \left(\frac{1}{5}\right)^2\right] = 0.72$$
$$\text{Gini(Money)} \rightarrow 0*\frac{5}{10} + 0.72*\frac{5}{10} = 0.36$$

````python
from math import log2
````

````python
def I(a,b):
    if a == 0 or b == 0:
        return 0
    one = a/(a+b)
    two = log2(one)
    first = one * two

    one = b/(a+b)
    two = log2(one)
    second = one * two
    return -first - second
````

````python
(5/14)*I(2,3)+(4/14)*I(4,0)+(5/14)*I(3,2)
````

````python
import numpy as np
import pandas as pd

veriler=pd.read_csv("veriler.csv")

x=veriler.iloc[:,1:4].values
y=veriler.iloc[:,4:].values

from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.33, random_state=0)
x_train
````

````python
from sklearn.tree import DecisionTreeClassifier
# default olarak gini kullanır, entropy'den farkı log2 tabanında almadan 
# çarpması yani karesini alması
dtc = DecisionTreeClassifier(criterion="entropy")
````

````python
dtc.fit(x_train,y_train)
y_pred = dtc.predict(x_test)
````

````python
from sklearn.metrics import confusion_matrix
cm = confusion_matrix(y_true=y_test, y_pred=y_pred)
cm
````
